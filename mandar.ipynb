decimo segundo

```python
def eh_primo(numero, divisor=2):
    if numero <= 1:
        return False
    if divisor > numero ** 0.5:
        return True
    if numero % divisor == 0:
        return False
    return eh_primo(numero, divisor + 1)

# Solicita um número ao usuário
num = int(input("Digite um número para verificar se é primo: "))

# Verifica se é primo e exibe o resultado
if eh_primo(num):
    print(f"{num} é um número primo.")
else:
    print(f"{num} não é um número primo.")
```
